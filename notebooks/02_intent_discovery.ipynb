```python
import pandas as pd
pd.set_option('display.max_colwidth', None)

# Notice the ../ at the start to go up one folder!
df = pd.read_csv("../data/processed/apple_support_conversations.csv")

# Print 20 random rows (just the customer text)
df['text_customer'].sample(20).tolist()
```
